## Load the experimental dataset and fitted results

```python
import csv
archive = np.load(SAVED_RESULTS / 'fits_best_available.npz', allow_pickle=True)
cdc_x, cper_x = archive['CDC_X'], archive['CPER_X']
targets = archive['targets'].tolist()
fits = archive['fits']
fit_species = np.array([fit['species'] for fit in fits])
with (ROOT / 'data/species_names.csv').open(newline='') as source:
    aliases = {row['species_code']: row['figure_code'] for row in csv.DictReader(source)}
species_labels = np.array([aliases.get(code, code) for code in fit_species])
```

## Extract species curves and parameters

```python
cdc_Y = []
cper_Y = []
cdc_Ym = []
cper_Ym = []
cdc_Ys = []
cper_Ys = []
losses = []
inh_mults = []
ada_mults = []
for fit in fits:
    cdc_target, cper_target = targets[fit['species']]
    cdc_Y.append(cdc_target)
    cper_Y.append(cper_target)
    cdc_Ym.append(fit['cdc'])
    cper_Ym.append(fit['cper'])
    cdc_Ys.append(fit['seed_cdc'])
    cper_Ys.append(fit['seed_cper'])
    losses.append(fit['loss'])
    inh_mults.append(fit['inh_mult'])
    ada_mults.append(fit['ada_mult'])
cdc_Y = np.array(cdc_Y)
cper_Y = np.array(cper_Y)
cdc_Ys = np.array(cdc_Ys)
cper_Ys = np.array(cper_Ys)
```

## Cluster the endpoint duty-cycle responses

```python
num_clusters = 3
Z = linkage(pdist(cdc_Y[:,-2:], metric="euclidean"), method="average", optimal_ordering=True,)
labels = fcluster(Z, t=num_clusters, criterion="maxclust") - 1
cut = (Z[-num_clusters, 2] + Z[-num_clusters + 1, 2]) / 2

tree = dendrogram(Z, color_threshold=cut, no_plot=True, orientation='left', labels=species_labels)
order = tree['leaves']
```

## Plot fits with correctly paired parameter bars

Gray curves show the experimental targets; black curves show the fitted model. Blue bars are the AN1 divisive-normalization multiplier, and red bars are the AN1 inhibitory-filter multiplier. All rows share the same parameter scale.

```python
from matplotlib.lines import Line2D

with plt.rc_context({'font.size': 10, 'axes.linewidth': .8, 'lines.linewidth': 1.2}):
    pub = plt.figure(figsize=(4.8, 596.227 / 72))
    bottom, span = .12, .82
    step = span / len(order)
    tree_ax = pub.add_axes([.01, bottom - .1 * step, .14, span])
    pub_tree = dendrogram(Z, color_threshold=cut, ax=tree_ax, orientation='left', no_labels=True)
    tree_ax.axis('off')
    pub.text(.19, .952, 'G', fontsize=28, weight='bold')
    pub.text(.82, .952, 'H', fontsize=28, weight='bold')
    for position, row in enumerate(order[::-1]):
        y = bottom + (len(order) - 1 - position) * step
        pub.text(.25, y + .4 * step, species_labels[row], ha='right', va='center', fontsize=10)
        for col, (xs, observed, predicted, xlabel) in enumerate([
            (cdc_x, cdc_Y[row], cdc_Ym[row], 'Duty cycle'),
            (cper_x / 1000, cper_Y[row], cper_Ym[row], 'Period [s]'),
        ]):
            ax = pub.add_axes([.27 + .27 * col, y, .21, .8 * step])
            ax.plot(xs, observed, color='0.6')
            ax.plot(xs, predicted, color='0.1')
            ax.set(xlim=(0, 1), ylim=(0, 1.08), yticks=[])
            ax.spines[['top', 'right', 'left']].set_visible(False)
            ax.set_xticks([])
            if position == len(order) - 1:
                ax.set_xticks([0, 1])
                ax.tick_params(length=2, pad=2, labelsize=10)
                ax.set_xlabel(xlabel, fontsize=10, labelpad=3)
        ax = pub.add_axes([.83, y, .16, .8 * step])
        bars = ax.bar([0, 1], [ada_mults[row], inh_mults[row]], width=.75,
                      color=['#3978ba', '#df4647'])
        ax.set(xlim=(-.65, 1.65), ylim=(0, 4), xticks=[], yticks=[])
        ax.spines[['top', 'right', 'left']].set_visible(False)
        if position == len(order) - 1:
            ax.spines['left'].set_visible(True)
            ax.set_yticks([0, 4])
            ax.set_xticks([0, 1], ['Div. norm.', 'Inh. comp.'], rotation=90, fontsize=9)
            ax.tick_params(length=2, pad=2, labelsize=9)
    pub.legend([Line2D([], [], color='0.1'), Line2D([], [], color='0.6')],
               ['Model', 'Data'], frameon=False, loc='lower left',
               bbox_to_anchor=(.005, .025), fontsize=9, handlelength=1.4)
    pub.savefig(FIGURES / 'fig_5G-H.pdf')
    plt.show()
```
